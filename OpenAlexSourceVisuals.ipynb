```python
def Matching_columns(df, col1, col2):
    match = df[col1].equals(df[col2])
    return match
```

```python
# primary topic
Matching_columns(df, 'primary_topic-display_name', 'topics-display_name')
```

```python
# subfield
Matching_columns(df, 'primary_topic-subfield', 'topics-subfield-id')
```

```python
# field
Matching_columns(df, 'primary_topic-field', 'topics-field-id')
```

```python
# domain
Matching_columns(df, 'primary_topic-domain', 'topics-domain-id')
```

```python
OpenAccessdf = df[df['is_oa'] == True]
OpenAccessdf = OpenAccessdf[['title', 'cited_by_count']].sort_values(by=['cited_by_count'], ascending=False).drop_duplicates().reset_index(drop=True)
print('Total open access cited by count is:', OpenAccessdf['cited_by_count'].sum())
print('Average open access cited by count is:', OpenAccessdf['cited_by_count'].mean())
OpenAccessdf.head(15)
```

```python
NonOpenAccessdf = df[df['is_oa'] == False]
NonOpenAccessdf = NonOpenAccessdf[['title', 'cited_by_count']].sort_values(by=['cited_by_count'], ascending=False).drop_duplicates().reset_index(drop=True)
print('Total non open access cited by count is:', NonOpenAccessdf['cited_by_count'].sum())
print('Average non open access cited by count is:', NonOpenAccessdf['cited_by_count'].mean())
NonOpenAccessdf.head(15)
```

```python
doi_count = df['doi'].nunique()

domain_cited_df = df.drop_duplicates(subset=['doi'])
domain_cited_df = domain_cited_df.groupby(['topics-domain-display_name'])['cited_by_count'].sum().reset_index().sort_values(by='cited_by_count', ascending=False).reset_index(drop=True)
#OpenAccessdf = OpenAccessdf[['title', 'cited_by_count']].sort_values(by=['cited_by_count'], ascending=False).drop_duplicates().reset_index(drop=True)
#print('total cited by count is:', OpenAccessdf['cited_by_count'].sum())
#domain['cited by mean'] = domain_cited_df.groupby(['topics-domain-display_name'])['cited_by_count'].sum().reset_index().sort_values(by='cited_by_count', ascending=False).reset_index(drop=True) / len 
domain_cited_df['cited_by_count_mean'] = domain_cited_df['cited_by_count'] / doi_count
domain_cited_df
```

```python
field_df = df.drop_duplicates(subset=['doi'])
field_df = field_df.groupby(['topics-field-display_name'])['cited_by_count'].sum().reset_index().sort_values(by='cited_by_count', ascending=False).reset_index(drop=True)
field_df['cited_by_count_mean'] = field_df['cited_by_count'] / doi_count

field_df.head(15)
```

```python
subfield_df = df
subfield_df = subfield_df.groupby(['topics-subfield-display_name'])['cited_by_count'].sum().reset_index().sort_values(by='cited_by_count', ascending=False).reset_index(drop=True)
subfield_df['cited_by_count_mean'] = subfield_df['cited_by_count'] / doi_count

subfield_df.head(15)
```

```python
topic_cited_df = df.drop_duplicates(subset=['doi'])
topic_cited_df = topic_cited_df.groupby(['primary_topic-display_name'])['cited_by_count'].sum().reset_index().sort_values(by='cited_by_count', ascending=False).reset_index(drop=True)
topic_cited_df['cited_by_count_mean'] = topic_cited_df['cited_by_count'] / doi_count


topic_cited_df.head(15)
```
