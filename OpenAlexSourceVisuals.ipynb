```python
def Matching_columns(df, col1, col2):
    match = df[col1].equals(df[col2])
    return match
```

```python
domain_df = df.drop_duplicates(subset=['doi']).groupby(['topics-domain-display_name']).agg(cited_by_count=('cited_by_count', 'sum'), dataset_count=('topics-domain-display_name', 'count')).reset_index()

domain_df['domain_dataset %'] = domain_df['dataset_count'] / domain_df['dataset_count'].sum()
domain_df['domain_citation_mean'] = domain_df['cited_by_count'] / domain_df['dataset_count']
domain_df = domain_df[['topics-domain-display_name', 'cited_by_count', 'domain_citation_mean', 'dataset_count', 'domain_dataset %']]

domain_df.sort_values(by='domain_citation_mean', ascending=False).reset_index(drop=True).head(15)
```

```python
field_df = df.drop_duplicates(subset=['doi']).groupby(['topics-field-display_name']).agg(cited_by_count=('cited_by_count', 'sum'), dataset_count=('topics-field-display_name', 'count')).reset_index()

field_df['field_dataset %'] = field_df['dataset_count'] / field_df['dataset_count'].sum()
field_df['field_citation_mean'] = field_df['cited_by_count'] / field_df['dataset_count']
field_df = field_df[['topics-field-display_name', 'cited_by_count', 'field_citation_mean', 'dataset_count', 'field_dataset %']]

field_df.sort_values(by='field_citation_mean', ascending=False).reset_index(drop=True).head(15)
```

```python
subfield_df = df.drop_duplicates(subset=['doi']).groupby(['topics-subfield-display_name']).agg(cited_by_count=('cited_by_count', 'sum'), dataset_count=('topics-subfield-display_name', 'count')).reset_index()

subfield_df['subfield_dataset %'] = subfield_df['dataset_count'] / subfield_df['dataset_count'].sum()
subfield_df['subfield_citation_mean'] = subfield_df['cited_by_count'] / subfield_df['dataset_count']
subfield_df = subfield_df[['topics-subfield-display_name', 'cited_by_count', 'subfield_citation_mean', 'dataset_count', 'subfield_dataset %']]

subfield_df.sort_values(by='subfield_citation_mean', ascending=False).reset_index(drop=True).head(15)
```
